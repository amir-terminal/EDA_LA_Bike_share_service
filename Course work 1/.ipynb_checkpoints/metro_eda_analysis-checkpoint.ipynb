## 1.2 Missing Data AssessmentData quality assessment begins with systematic analysis of missing values. The dataset exhibits localized missing values primarily in geographic coordinates, which is common in GPS-based systems. Understanding missingness patterns is essential for appropriate handling strategies.

```python
print()print('MISSING DATA ANALYSIS')print('='*80)missing = df.isnull().sum()missing_pct = (missing / len(df) * 100).round(2)miss_df = pd.DataFrame({    'Column': df.columns,    'Missing_Count': missing.values,    'Missing_Pct': missing_pct.values})print(miss_df[miss_df['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False))print()total_incomplete = df.isnull().any(axis=1).sum()print(f'Total records with missing values: {total_incomplete} ({total_incomplete/len(df)*100:.2f}%)')
```

**Key Finding:** Approximately 4.42% of records have missing end coordinates and 2.32% have missing start coordinates. These missing values are concentrated in geographic fields, suggesting systematic GPS signal loss during certain trips rather than random data collection failures. For this analysis, records with missing geographic data are retained for non-spatial analyses (e.g., temporal patterns by hour/day, user segmentation) while being excluded from geospatial studies. This approach maximizes usable information while maintaining analytical integrity.

```python
print()print('CATEGORICAL VARIABLES')print('='*80)print('Trip Route Category:')print(df['trip_route_category'].value_counts())print()print('Passholder Type:')ph = df['passholder_type'].value_counts()for pt, count in ph.items():    print(f'  {pt}: {count:,} ({count/len(df)*100:.2f}%)')print()print('Bike Type:')for bt, count in df['bike_type'].value_counts().items():    print(f'  {bt}: {count:,} ({count/len(df)*100:.2f}%)')
```

# 2. Data Cleansing and Preprocessing## 2.1 Outlier Detection and AnalysisOutlier analysis is essential for understanding data quality and identifying exceptional observations. Trip duration outliers may represent system malfunctions (under-counted trips) or legitimate extended journeys. Using the Interquartile Range (IQR) method, we identify and characterize outliers.

```python
print()print('OUTLIER DETECTION ANALYSIS')print('='*80)Q1 = df['duration'].quantile(0.25)Q3 = df['duration'].quantile(0.75)IQR = Q3 - Q1lower_bound = Q1 - 1.5 * IQRupper_bound = Q3 + 1.5 * IQRprint(f'Quartile Analysis:')print(f'  Q1 (25th percentile): {Q1:.1f} minutes')print(f'  Q3 (75th percentile): {Q3:.1f} minutes')print(f'  IQR: {IQR:.1f} minutes')print()print(f'IQR Method Bounds:')print(f'  Lower bound: {lower_bound:.1f} minutes')print(f'  Upper bound: {upper_bound:.1f} minutes')outliers = df[(df['duration'] < lower_bound) | (df['duration'] > upper_bound)]print()print(f'Outliers identified: {len(outliers):,} records ({len(outliers)/len(df)*100:.2f}%)')print()print(f'Outlier duration statistics:')print(f'  Min: {outliers["duration"].min()} minutes')print(f'  Max: {outliers["duration"].max()} minutes')print(f'  Mean: {outliers["duration"].mean():.1f} minutes')print(f'  Median: {outliers["duration"].median():.1f} minutes')
```

**Interpretation of Outliers:** The IQR method identifies 8,291 outliers (9.0% of data), representing trips exceeding 39 minutes. These outliers are not errors but legitimate extended journeys, predominantly from recreational users (Walk-up, One Day Pass holders). Rather than removal, outliers are retained to preserve information about recreational usage patterns. The distribution of outliers skews heavily toward longer durations (max 1440 minutes = 24 hours), suggesting users leaving bikes out overnight or system calibration issues.

## 2.2 Data Preprocessing and Feature EngineeringPreprocessing transforms raw data into analytical-ready format. Key preprocessing steps include datetime parsing and temporal feature extraction, enabling analysis of usage patterns by time of day and day of week.

```python
# Parse datetime fieldsdf['start_time'] = pd.to_datetime(df['start_time'])df['end_time'] = pd.to_datetime(df['end_time'])# Extract temporal featuresdf['start_hour'] = df['start_time'].dt.hourdf['start_day_of_week'] = df['start_time'].dt.day_name()df['start_date'] = df['start_time'].dt.datedf['start_month'] = df['start_time'].dt.monthdf['trip_hour_duration'] = (df['end_time'] - df['start_time']).dt.total_seconds() / 60# Verify duration consistencyduration_mismatch = (df['duration'] != df['trip_hour_duration']).sum()print(f'Duration field verification: {duration_mismatch} discrepancies found')# Standardize day of week orderingday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']df['start_day_of_week'] = pd.Categorical(df['start_day_of_week'],                                           categories=day_order, ordered=True)print()print('Temporal Features Created:')print(f'  Date range: {df["start_date"].min()} to {df["start_date"].max()}')print(f'  Span: {(df["start_date"].max() - df["start_date"].min()).days} days')print(f'  Hour range: 0-{df["start_hour"].max()}')print()print('Sample processed data:')print(df[['trip_id', 'start_time', 'start_hour', 'start_day_of_week', 'duration']].head())
```

# 3. Exploratory Data Visualization and Insights## 3.1 Duration Distribution and CharacteristicsUnderstanding the distribution of trip duration is fundamental to user behavior analysis. The distribution shape, central tendency, and spread reveal underlying usage patterns.

```python
# Distribution analysisduration_0_120 = df[df['duration'] <= 120]['duration']print()print('TRIP DURATION DISTRIBUTION ANALYSIS')print('='*80)print()print(f'Full Dataset (all durations):')print(f'  Count: {len(df):,}')print(f'  Mean: {df["duration"].mean():.2f} minutes')print(f'  Median: {df["duration"].median():.2f} minutes')print(f'  Std Dev: {df["duration"].std():.2f} minutes')print(f'  Skewness: {df["duration"].skew():.3f}')print(f'  Kurtosis: {df["duration"].kurtosis():.3f}')print()print(f'Filtered Dataset (0-120 minutes):')print(f'  Count: {len(duration_0_120):,} ({len(duration_0_120)/len(df)*100:.1f}%)')print(f'  Mean: {duration_0_120.mean():.2f} minutes')print(f'  Median: {duration_0_120.median():.2f} minutes')print(f'  Std Dev: {duration_0_120.std():.2f} minutes')print(f'  Skewness: {duration_0_120.skew():.3f}')print(f'  Kurtosis: {duration_0_120.kurtosis():.3f}')print()print('Percentiles:')for pct in [10, 25, 50, 75, 90, 95, 99]:    val = df['duration'].quantile(pct/100)    print(f'  {pct}th: {val:.1f} minutes')
```

## 3.2 Duration by Passholder Type: Statistical AnalysisUser segmentation by passholder type reveals distinct usage behaviors. Subscription users (Monthly, Annual Pass) and pay-as-you-go users (Walk-up, One Day Pass) show dramatically different duration patterns, suggesting fundamentally different use cases.

```python
print()print('DURATION BY PASSHOLDER TYPE')print('='*80)print()duration_by_ph = df.groupby('passholder_type')['duration'].agg([    'count', 'mean', 'median', 'std', 'min', 'max']).round(2)duration_by_ph = duration_by_ph.sort_values('mean', ascending=False)print(duration_by_ph)print()print('Interpretation:')print('  - Walk-up users: Mean 69.7 min (median 22 min) - recreational/leisure usage')print('  - One Day Pass: Mean 79.5 min (median 27 min) - full-day passes enable longer trips')print('  - Monthly/Annual: Mean ~15-16 min (median 8-9 min) - commuter usage patterns')
```

### Statistical Hypothesis Testing: ANOVATo statistically confirm differences in duration across passholder types, we conduct Analysis of Variance (ANOVA). The null hypothesis posits equal mean durations across groups; the alternative hypothesis suggests at least one group differs.

```python
# ANOVA test for duration by passholder typegroups = [group['duration'].values for name, group in df.groupby('passholder_type')]f_stat, p_value = stats.f_oneway(*groups)print()print('ANOVA TEST: Duration ~ Passholder Type')print('='*80)print(f'Null Hypothesis: All passholder type groups have equal mean duration')print(f'Alternative: At least one group differs')print()print(f'F-statistic: {f_stat:.4f}')print(f'P-value: {p_value:.2e}')print()if p_value < 0.001:    print('RESULT: Reject null hypothesis (p < 0.001)')    print('CONCLUSION: Significant differences in mean trip duration exist across passholder types.')else:    print('RESULT: Fail to reject null hypothesis')print()print('Effect Size Interpretation:')print('  The extremely small p-value indicates highly significant differences.')print('  Passholder type is a strong predictor of trip duration.')
```

## 3.3 Temporal Patterns: Hour of Day and Day of WeekTemporal analysis reveals when and how users engage with the bike share system. Distinct patterns emerge between commuting (weekday morning/evening peaks) and recreational usage (weekend activity).

```python
print()print('DURATION BY HOUR OF DAY')print('='*80)print()duration_by_hour = df.groupby('start_hour')['duration'].agg([    'count', 'mean', 'median', 'std']).round(2)print(duration_by_hour)print()print()print('DURATION BY DAY OF WEEK')print('='*80)print()duration_by_day = df.groupby('start_day_of_week')['duration'].agg([    'count', 'mean', 'median', 'std']).round(2)print(duration_by_day)print()print('Key Temporal Insights:')print('  - Late night (0-4am): Mean 57-105 min - recreational evening/night rides')print('  - Morning commute (6-8am): Mean 15-17 min - rapid point-to-point commuting')print('  - Midday (9-17): Mean 27-37 min - mixed commute and leisure')print('  - Weekdays (Mon-Fri): Mean 27-32 min - commuter-dominated')print('  - Weekends (Sat-Sun): Mean 44-46 min - 60% longer, recreational usage')
```

### Statistical Testing: Kruskal-Wallis Test for Day of WeekThe Kruskal-Wallis test is a non-parametric alternative to ANOVA, appropriate given the non-normal distribution of duration data (skewness: 2.646, kurtosis: 8.771).

```python
# Kruskal-Wallis test for duration by day of weekgroups_day = [group['duration'].values for name, group in df.groupby('start_day_of_week')]h_stat, p_day = stats.kruskal(*groups_day)print()print('KRUSKAL-WALLIS TEST: Duration ~ Day of Week')print('='*80)print(f'H-statistic: {h_stat:.4f}')print(f'P-value: {p_day:.2e}')print()if p_day < 0.001:    print('RESULT: Reject null hypothesis (p < 0.001)')    print('CONCLUSION: Significant differences in trip duration across days of week.')else:    print('RESULT: Fail to reject null hypothesis')print()print('Practical Interpretation:')print('  Weekend riders take substantially longer trips (recreational use).')print('  Weekday riders take shorter, more direct trips (commuting).')
```

## 3.4 Impact of Bike Type on Trip DurationDifferent bike types (standard, electric, smart) serve different user needs. Electric bikes reduce physical effort, enabling longer journeys, while smart bikes provide user guidance.

```python
print()print('DURATION BY BIKE TYPE')print('='*80)print()duration_by_bike = df.groupby('bike_type')['duration'].agg([    'count', 'mean', 'median', 'std']).round(2)duration_by_bike = duration_by_bike.sort_values('mean', ascending=False)print(duration_by_bike)print()print('Interpretation:')print('  - Smart bikes: Mean 45.5 min - longest duration, newer technology')print('  - Electric bikes: Mean 30.6 min - mid-range, assist enables longer rides')print('  - Standard bikes: Mean 30.5 min - pedal-only, shortest duration')print()print('Analysis:')print('  Smart bikes support ~50% longer trips than standard bikes.')print('  Electric vs Standard show minimal difference (30.6 vs 30.5 min).')print('  Bike type explains variation in user trip patterns.')
```

## 3.5 Trip Route CharacteristicsThe distinction between one-way and round-trip journeys provides insight into trip planning and user intention.

```python
print()print('DURATION BY TRIP ROUTE CATEGORY')print('='*80)print()duration_by_route = df.groupby('trip_route_category')['duration'].agg([    'count', 'mean', 'median', 'std']).round(2)print(duration_by_route)print()print('Interpretation:')one_way_mean = duration_by_route.loc['One Way', 'mean']round_trip_mean = duration_by_route.loc['Round Trip', 'mean']pct_diff = ((round_trip_mean - one_way_mean) / one_way_mean * 100)print(f'  - One Way: Mean {one_way_mean:.1f} minutes, {duration_by_route.loc["One Way", "count"]:.0f} trips (83.4%)')print(f'  - Round Trip: Mean {round_trip_mean:.1f} minutes, {duration_by_route.loc["Round Trip", "count"]:.0f} trips (16.7%)')print()print(f'Round-trip duration is {pct_diff:.1f}% longer than one-way.')print('This suggests users plan longer, more leisurely round-trip routes.')
```
